##### Armstrong Number Checker
##### Problem Statement:
Check whether a given number is an Armstrong number. A number is an Armstrong number if it is equal to the sum of the cubes of its digits.
##### Input:
- A positive integer
##### Output:
- YES / NO
##### Description
This program checks whether a number is an Armstrong number by extracting each digit, calculating the cube of the digit, and adding the cubes together. If the sum equals the original number, the program prints **YES**; otherwise, it prints **NO**.

```python
# Step 1: Take input from the user
number = int(input("Enter a number: "))
print(f"Entered Number: {number}")

# Step 2: Store the original number
original_number = number

# Step 3: Calculate the sum of cubes of digits
sum_of_cubes = 0

while number > 0:
    digit = number % 10
    sum_of_cubes += digit ** 3
    number = number // 10

# Step 4: Check whether the number is an Armstrong number
if sum_of_cubes == original_number:
    print("YES")
else:
    print("NO")
```
