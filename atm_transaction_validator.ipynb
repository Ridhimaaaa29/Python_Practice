##### ATM Transaction Validator
Problem Statement: An ATM processes **N** withdrawal requests sequentially. Each request has an amount.
##### Rules:
- Withdrawal amount must be a **multiple of 100**.
- Account balance must **never become negative**.
- For each transaction, print **SUCCESS** or **FAILED**.
- Display the **final account balance** after processing all transactions.
##### Input:
- Initial account balance
- Number of transactions (N)
- Withdrawal amount for each transaction
##### Output:
- SUCCESS or FAILED for each transaction
- Final account balance
##### Description
This program simulates an ATM transaction system. It validates each withdrawal request based on two conditions: whether the withdrawal amount is a multiple of 100 and whether sufficient balance is available. If both conditions are satisfied, the amount is deducted from the balance; otherwise, the transaction is rejected. After all transactions are processed, the remaining account balance is displayed.

```python
# Step 1: Take initial account balance
account_balance = int(input("Enter the initial account balance: "))

# Step 2: Take the number of withdrawal requests
number_of_transactions = int(input("Enter the number of transactions: "))

# Step 3: Process each withdrawal request
for transaction in range(1, number_of_transactions + 1):

    withdrawal_amount = int(input(f"Enter withdrawal amount {transaction}: "))

    # Check if amount is a multiple of 100 and balance is sufficient
    if withdrawal_amount % 100 == 0 and withdrawal_amount <= account_balance:
        account_balance -= withdrawal_amount
        print("SUCCESS")
    else:
        print("FAILED")

# Step 4: Display the final account balance
print(f"Final Account Balance: {account_balance}")
```
