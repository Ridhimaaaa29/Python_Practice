##### Binary to Decimal Converter (Without Built-in)
##### Problem Statement:
Convert a given binary number into its decimal equivalent without using any built-in conversion functions.
##### Input:
- Binary number
##### Output:
- Decimal number
##### Description
This program converts a binary number into its decimal equivalent by processing each binary digit from right to left. It uses powers of 2, arithmetic operations, loops, and conditional statements instead of Python's built-in conversion functions.

```python
# Step 1: Take binary number input from the user
binary_number = input("Enter the binary number: ")
print(f"Binary Number: {binary_number}")

# Step 2: Initialize variables
decimal_number = 0
power = 0

# Step 3: Process digits from right to left
for digit in binary_number[::-1]:

    if digit == "1":
        decimal_number += 2 ** power

    power += 1

# Step 4: Display the decimal equivalent
print(f"Decimal Number: {decimal_number}")
```
