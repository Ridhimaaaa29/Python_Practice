##### Bus Seat Allocation
##### Problem Statement:
A bus has **40 seats**. For each booking request, allocate seats based on availability.
##### Rules:
- If enough seats are available, print **CONFIRMED**.
- Otherwise, print **WAITLISTED**.
##### Input:
- Number of booking requests (N)
- Seats requested for each booking
##### Output:
- CONFIRMED / WAITLISTED for each booking request
##### Description
This program simulates a bus seat reservation system with a total of 40 seats. For each booking request, it checks whether the required seats are available. If sufficient seats remain, the booking is confirmed and the available seats are updated; otherwise, the request is waitlisted.

```python
# Step 1: Initialize total seats
available_seats = 40

# Step 2: Take number of booking requests
number_of_requests = int(input("Enter the number of booking requests: "))

# Step 3: Process each booking request
for request in range(1, number_of_requests + 1):

    seats_requested = int(input(f"Enter seats requested for booking {request}: "))

    print(f"Booking {request}: Requested {seats_requested} seat(s)")

    # Step 4: Check seat availability
    if seats_requested <= available_seats:
        print("CONFIRMED")
        available_seats -= seats_requested
    else:
        print("WAITLISTED")

# Step 5: Display remaining seats
print(f"\nRemaining Seats: {available_seats}")
```
