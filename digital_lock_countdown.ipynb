##### Digital Lock Countdown
##### Problem Statement:
Given a number, repeatedly calculate the sum of its digits until the result becomes a single-digit number.
##### Input:
- A positive integer
##### Output:
- Final single-digit number
##### Description
This program repeatedly finds the sum of the digits of a number until only a single-digit value remains. It uses nested loops to calculate the digit sum and reduce the number step by step.

```python
# Step 1: Take input from the user
number = int(input("Enter a number: "))
print(f"Entered Number: {number}")

# Step 2: Repeat until the number becomes a single digit
while number >= 10:

    digit_sum = 0

    # Step 3: Calculate the sum of digits
    while number > 0:
        digit = number % 10
        digit_sum += digit
        number = number // 10

    # Step 4: Update the number with the digit sum
    number = digit_sum

# Step 5: Display the result
print(f"Final Single Digit: {number}")
```
