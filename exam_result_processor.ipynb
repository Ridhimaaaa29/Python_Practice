##### Exam Result Processor
##### Problem Statement:
Input marks for **5 subjects** and determine the student's result based on the following rules.
##### Rules:
- If **any subject mark is less than 35**, print **FAIL**.
- Otherwise, if the **average marks are 75 or above**, print **DISTINCTION**.
- Otherwise, print **PASS**.
##### Input:
- Marks of 5 subjects
##### Output:
- FAIL / PASS / DISTINCTION
##### Description
This program evaluates a student's exam result by first checking whether the student has failed in any subject. If the student passes all subjects, it calculates the average marks and classifies the result as either **DISTINCTION** or **PASS**.

```python
# Step 1: Take marks of five subjects as input
mark1 = float(input("Enter marks of Subject 1: "))
mark2 = float(input("Enter marks of Subject 2: "))
mark3 = float(input("Enter marks of Subject 3: "))
mark4 = float(input("Enter marks of Subject 4: "))
mark5 = float(input("Enter marks of Subject 5: "))

# Display the entered marks
print(f"\nMarks Entered: {mark1}, {mark2}, {mark3}, {mark4}, {mark5}")

# Step 2: Check if the student has failed in any subject
if (mark1 < 35 or mark2 < 35 or mark3 < 35 or mark4 < 35 or mark5 < 35):
    print("Result: FAIL")

else:
    # Step 3: Calculate average marks
    average_marks = (mark1 + mark2 + mark3 + mark4 + mark5) / 5

    # Step 4: Determine the result
    if average_marks >= 75:
        print("Result: DISTINCTION")
    else:
        print("Result: PASS")
```
