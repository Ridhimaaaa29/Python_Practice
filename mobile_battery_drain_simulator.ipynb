##### Mobile Battery Drain Simulator
##### Problem Statement:
A mobile battery starts at **100%**. An application drains a fixed percentage of battery every minute. Continue draining the battery until it becomes **0% or less**, then display the total number of minutes the battery lasted.
##### Input:
- Battery drain percentage per minute
##### Output:
- Total minutes until the battery is completely drained
##### Description
This program simulates battery consumption over time. It repeatedly reduces the battery percentage by a fixed amount every minute until the battery is fully drained. Finally, it displays the total number of minutes the battery lasted.

```python
# Step 1: Take battery drain percentage per minute
drain_per_minute = float(input("Enter the battery drain percentage per minute: "))
print(f"Battery Drain per Minute: {drain_per_minute}%")

# Step 2: Initialize battery percentage and minute counter
battery_percentage = 100
minutes = 0

# Step 3: Drain battery until it becomes 0 or less
while battery_percentage > 0:
    battery_percentage -= drain_per_minute
    minutes += 1

# Step 4: Display the result
print(f"Minutes Used: {minutes}")
```
