##### Number Compression Counter
##### Problem Statement:
Given a number, count how many times it can be divided by **2** until it becomes an odd number.
##### Input:
- A positive integer
##### Output:
- Number of times the value can be divided by 2
##### Description
This program repeatedly divides a number by **2** as long as it remains even. It counts the number of successful divisions and displays the final count.

```python
# Step 1: Take input from the user
number = int(input("Enter a number: "))
print(f"Entered Number: {number}")

# Step 2: Initialize counter
count = 0

# Step 3: Divide the number by 2 until it becomes odd
while number % 2 == 0:
    number = number // 2
    count += 1

# Step 4: Display the result
print(f"Number of divisions by 2: {count}")
```
