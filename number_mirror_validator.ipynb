##### Number Mirror Validator
##### Problem Statement:
Reverse a given number and determine whether the original number is equal to its reverse.
##### Input:
- A positive integer
##### Output:
- PALINDROME / NOT PALINDROME
##### Description
This program reverses a number using arithmetic operations and compares the reversed number with the original number. If both numbers are equal, the number is a palindrome; otherwise, it is not.

```python
# Step 1: Take input from the user
number = int(input("Enter a number: "))
print(f"Entered Number: {number}")

# Step 2: Store the original number
original_number = number

# Step 3: Reverse the number
reversed_number = 0

while number > 0:
    digit = number % 10
    reversed_number = (reversed_number * 10) + digit
    number = number // 10

# Step 4: Check if the number is a palindrome
if original_number == reversed_number:
    print("PALINDROME")
else:
    print("NOT PALINDROME")
```
