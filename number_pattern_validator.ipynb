##### Number Pattern Validator
##### Problem Statement:
Check whether the digits of a given number are **strictly increasing** from left to right.
##### Input:
- A positive integer
##### Output:
- YES / NO
##### Description
This program checks whether each digit of a number is greater than the digit before it. It compares adjacent digits one by one and determines whether the entire number follows a strictly increasing pattern.

```python
# Step 1: Take input from the user
number = int(input("Enter a number: "))
print(f"Entered Number: {number}")

# Step 2: Initialize variables
is_increasing = True
previous_digit = 10

# Step 3: Compare adjacent digits
while number > 0:

    current_digit = number % 10

    if current_digit >= previous_digit:
        is_increasing = False
        break

    previous_digit = current_digit
    number = number // 10

# Step 4: Display the result
if is_increasing:
    print("YES")
else:
    print("NO")
```
