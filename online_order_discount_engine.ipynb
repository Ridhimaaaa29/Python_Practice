##### Online Order Discount Engine
##### Problem Statement:
Calculate the final payable amount after applying the highest eligible discount based on the total order amount.
##### Rules:
- Order amount ≥ 5000 → 20% discount
- Order amount ≥ 3000 → 10% discount
- Order amount ≥ 1000 → 5% discount
- Otherwise → No discount
- Apply only one highest applicable discount.
##### Input:
- Total order amount
##### Output:
- Final payable amount
##### Description
This program calculates the final payable amount after applying the highest eligible discount. It demonstrates conditional statements, percentage calculations, arithmetic operations, and decision-making using if-elif-else.

```python
# Step 1: Take input from the user
order_amount = float(input("Enter the total order amount: "))
print(f"Order Amount: ₹{order_amount:.2f}")

# Step 2: Calculate discount percentage
if order_amount >= 5000:
    discount_percentage = 20
elif order_amount >= 3000:
    discount_percentage = 10
elif order_amount >= 1000:
    discount_percentage = 5
else:
    discount_percentage = 0

# Step 3: Calculate payable amount
discount_amount = order_amount * discount_percentage / 100
payable_amount = order_amount - discount_amount

# Step 4: Display the result
print(f"Payable Amount: ₹{payable_amount:.2f}")
```
