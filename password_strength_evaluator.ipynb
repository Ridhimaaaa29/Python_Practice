##### **Password Strength Evaluator**
##### Problem Statement:
Given a password string, determine whether it is **STRONG** or **WEAK** based on the following conditions.
##### Rules:
- Password must contain at least **1 digit**.
- Password must contain at least **1 uppercase letter**.
- Password length must be **greater than or equal to 8** characters.
##### Input:
- Password string
##### Output:
- STRONG if all conditions are satisfied.
- WEAK otherwise.
##### Description
This program evaluates the strength of a password by checking its length, the presence of at least one uppercase letter, and at least one numeric digit. It demonstrates string traversal, conditional statements, character validation, and logical operators to classify a password as STRONG or WEAK.

```python
# Step 1: Take password input from the user
password = input("Enter the password: ")

# Display the entered password
print(f"Entered Password: {password}")

# Step 2: Initialize counters
uppercase_count = 0
digit_count = 0

# Step 3: Loop through each character
for character in password:

    if character.isupper():
        uppercase_count += 1

    if character.isdigit():
        digit_count += 1

# Step 4: Check password strength
if len(password) >= 8 and uppercase_count >= 1 and digit_count >= 1:
    print("STRONG")
else:
    print("WEAK")
```
