##### Prime Range Analyzer
##### Problem Statement:
Find the total number of prime numbers between two given numbers (inclusive).
##### Rules:
- Check every number between A and B.
- Count only prime numbers.
- Check divisibility only up to the square root of the number.
##### Input:
- Starting number (A)
- Ending number (B)
##### Output:
- Total count of prime numbers
##### Description
This program counts the number of prime numbers within a given range. It demonstrates nested loops, conditional statements, efficient prime number checking, and mathematical optimization using the square root concept.

```python
# Step 1: Take input from the user
start_number = int(input("Enter the starting number: "))
print(f"Starting Number: {start_number}")

end_number = int(input("Enter the ending number: "))
print(f"Ending Number: {end_number}")

# Step 2: Initialize prime count
prime_count = 0

# Step 3: Check every number in the range
for number in range(start_number, end_number + 1):

    if number < 2:
        continue

    is_prime = True

    # Step 4: Check divisibility up to square root
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            is_prime = False
            break

    if is_prime:
        prime_count += 1

# Step 5: Display the result
print(f"Prime Count: {prime_count}")
```
