##### Salary Deduction System
##### Problem Statement:
Calculate the final salary of an employee after applying deductions based on late days and absent days.
##### Rules:
- Basic salary is provided.
- If late days are greater than 5, deduct **5%** of the salary.
- If late days are greater than 10, deduct **10%** of the salary.
- If absent days are greater than 2, deduct an **additional 5%** of the salary.
- Apply deductions cumulatively.
##### Input:
- Basic salary
- Number of late days
- Number of absent days
##### Output:
- Final salary after deductions
##### Description
This program calculates an employee's final salary after applying deductions based on attendance. It demonstrates the use of conditional statements, arithmetic operations, cumulative deductions, and percentage calculations.

```python
# Step 1: Take input from the user
salary = float(input("Enter the basic salary: "))
print(f"Basic Salary: {salary}")

late_days = int(input("Enter the number of late days: "))
print(f"Late Days: {late_days}")

absent_days = int(input("Enter the number of absent days: "))
print(f"Absent Days: {absent_days}")

# Step 2: Initialize deduction percentage
deduction_percentage = 0

# Step 3: Calculate deduction based on late days
if late_days > 10:
    deduction_percentage += 10
elif late_days > 5:
    deduction_percentage += 5

# Step 4: Add deduction for absent days
if absent_days > 2:
    deduction_percentage += 5

# Step 5: Calculate final salary
final_salary = salary - (salary * deduction_percentage / 100)

# Step 6: Display the result
print(f"Final Salary: {final_salary:.2f}")
```
