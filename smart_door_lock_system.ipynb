##### Smart Door Lock System
##### Problem Statement:
A user gets **3 attempts** to enter the correct PIN.
##### Rules:
- If the correct PIN is entered within 3 attempts, print **ACCESS GRANTED**.
- If all 3 attempts are incorrect, print **LOCKED**.
##### Input:
- Correct PIN
- Attempt 1
- Attempt 2
- Attempt 3
##### Output:
- ACCESS GRANTED / LOCKED
##### Description
This program simulates a smart door lock system by allowing the user up to three attempts to enter the correct PIN. The program grants access immediately when the correct PIN is entered and stops checking further attempts. If all three attempts are incorrect, the door remains locked.

```python
# Step 1: Take the correct PIN as input
correct_pin = input("Enter the correct PIN: ")

# Step 2: Initialize access status
access_granted = False

# Step 3: Allow the user up to 3 attempts
for attempt in range(1, 4):

    entered_pin = input(f"Enter PIN for attempt {attempt}: ")

    print(f"Attempt {attempt}: PIN Entered = {entered_pin}")

    # Step 4: Check whether the entered PIN is correct
    if entered_pin == correct_pin:
        print("ACCESS GRANTED")
        access_granted = True
        break

# Step 5: Lock the system if all attempts fail
if not access_granted:
    print("LOCKED")
```
