##### Smart Electricity Billing
##### Problem Statement:
Electricity bill is calculated slab-wise.
##### Rules:
- First **100 units** → ₹3 per unit.
- Next **100 units** → ₹5 per unit.
- Remaining units → ₹8 per unit.
- If the total usage is **more than 300 units**, add a **10% surcharge** to the total bill.
##### Input:
- Total electricity units consumed.
##### Output:
- Total electricity bill.
##### Description
This program calculates the electricity bill based on slab-wise pricing. It applies different rates for different ranges of electricity consumption and adds a 10% surcharge if the total consumption exceeds 300 units. Finally, it displays the total bill amount.

```python
# Step 1: Take input for electricity units consumed
units = int(input("Enter the total electricity units consumed: "))
print(f"Electricity Units Consumed: {units}")

# Step 2: Calculate the bill based on slab-wise pricing
if units <= 100:
    total_bill = units * 3

elif units <= 200:
    total_bill = (100 * 3) + ((units - 100) * 5)

else:
    total_bill = (100 * 3) + (100 * 5) + ((units - 200) * 8)

# Step 3: Apply 10% surcharge if units exceed 300
if units > 300:
    total_bill = total_bill + (total_bill * 0.10)

# Step 4: Display the final bill
print(f"Total Electricity Bill: ₹{total_bill:.2f}")
```
