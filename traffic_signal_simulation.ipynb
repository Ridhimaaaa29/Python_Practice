##### Traffic Signal Simulation
##### Problem Statement:
A traffic signal changes its color every second according to the following cycle.
##### Rules:
- Seconds **1–30** → RED
- Seconds **31–45** → YELLOW
- Seconds **46–90** → GREEN
##### Input:
- Time (T) in seconds
##### Output:
- RED, YELLOW, or GREEN
##### Description
This program simulates a traffic signal based on the given time in seconds. It uses conditional statements and range checking to determine the current signal color. The program also demonstrates the use of modulo arithmetic to handle repeating signal cycles.

```python
# Step 1: Take time input from the user
time_in_seconds = int(input("Enter the time in seconds: "))
print(f"Time in Seconds: {time_in_seconds}")

# Step 2: Repeat the signal cycle using modulo operator
signal_time = time_in_seconds % 90

# Step 3: Handle the case when the remainder is 0
if signal_time == 0:
    signal_time = 90

# Step 4: Determine the signal color
if 1 <= signal_time <= 30:
    print("RED")
elif 31 <= signal_time <= 45:
    print("YELLOW")
else:
    print("GREEN")
```
