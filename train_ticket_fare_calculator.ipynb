##### Train Ticket Fare Calculator
##### Problem Statement:
Calculate the train ticket fare based on the travel distance and passenger's age.
##### Rules:
- Fare = **₹2 per km**
- Senior Citizen (Age ≥ 60) → **30% discount**
- Child (Age < 12) → **50% discount**
- Others → **No discount**
##### Input:
- Travel distance (in km)
- Passenger age
##### Output:
- Final ticket fare
##### Description
This program calculates the base train fare using the travel distance and then applies an age-based discount. Senior citizens receive a 30% discount, children below 12 years receive a 50% discount, and all other passengers pay the full fare.

```python
# Step 1: Take input from the user
distance = float(input("Enter the travel distance (in km): "))
age = int(input("Enter the passenger's age: "))

# Display the entered inputs
print(f"\nTravel Distance: {distance} km")
print(f"Passenger Age: {age}")

# Step 2: Calculate the base fare
fare = distance * 2

# Step 3: Apply age-based discount
if age >= 60:
    fare = fare - (fare * 30 / 100)

elif age < 12:
    fare = fare - (fare * 50 / 100)

# Step 4: Display the final fare
print(f"Final Ticket Fare: ₹{fare:.2f}")
```
