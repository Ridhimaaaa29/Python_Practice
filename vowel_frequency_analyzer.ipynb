##### Vowel Frequency Analyzer
##### Problem Statement:
Count the total number of vowels in a given sentence (case-insensitive).
##### Input:
- A sentence
##### Output:
- Total number of vowels
##### Description
This program counts the total number of vowels present in a sentence. It first converts the sentence to lowercase so that uppercase and lowercase vowels are treated equally, then iterates through each character and counts the vowels.

```python
# Step 1: Take input from the user
sentence = input("Enter a sentence: ")
print(f"Entered Sentence: {sentence}")

# Step 2: Convert the sentence to lowercase
sentence = sentence.lower()

# Step 3: Initialize vowel counter
vowel_count = 0

# Step 4: Count vowels
for character in sentence:
    if character in "aeiou":
        vowel_count += 1

# Step 5: Display the result
print(f"Total Vowels: {vowel_count}")
```
