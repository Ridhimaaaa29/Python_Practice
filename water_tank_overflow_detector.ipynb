##### Water Tank Overflow Detector
##### Problem Statement:
A water tank has a capacity of **1000 liters**. Water flows into the tank every minute. Determine the minute at which the tank overflows.
##### Input:
- Number of minutes (N)
- Water inflow for each minute
##### Output:
- Minute at which the tank overflows
##### Description
This program simulates filling a water tank by adding the water inflow for each minute. It keeps track of the total water collected and stops as soon as the tank capacity exceeds **1000 liters**, displaying the minute when the overflow occurs.

```python
# Step 1: Define tank capacity
tank_capacity = 1000

# Step 2: Take the number of minutes
number_of_minutes = int(input("Enter the number of minutes: "))

# Step 3: Initialize current water level
current_water = 0

# Step 4: Take inflow for each minute
for minute in range(1, number_of_minutes + 1):

    inflow = int(input(f"Enter water inflow for minute {minute}: "))

    print(f"Minute {minute}: Inflow = {inflow} liters")

    current_water += inflow

    # Step 5: Check for overflow
    if current_water > tank_capacity:
        print(f"Overflow occurred at minute {minute}")
        break
```
